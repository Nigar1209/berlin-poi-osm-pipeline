## Objective
Analyze all **Point of Interest (POI)** tables in the `berlin_source_data` schema of the **production database** to ensure that `district_id` and `neighborhood_id` columns exist, have proper constraints, and contain valid data. The goal is to maintain referential integrity, prevent orphan records, and ensure consistent spatial and relational data across the platform.

This task focuses on validating:

- Presence of `district_id` and `neighborhood_id` columns where required  
- Proper constraints, including foreign keys and NOT NULL constraints  
- Correct assignment of IDs  
- Logical coverage of districts in POI tables  
- Detection of missing, orphan, or invalid IDs
