```python
import boto3
import os

# --- 1. CONFIGURATION ---

# ⚠️ REPLACE with the profile name you configured using 'aws configure'
# Example profile name from your previous output: 'AdministratorAccess-489962060388'
PROFILE_NAME = 'your_profile_name' 

# The name of your S3 bucket
BUCKET_NAME = 'your_bucket_name'

# Local path to your DAG file
LOCAL_FILE_PATH = 'path_to_your_file'

# The path/key in S3 where the file will be stored
S3_KEY = 'dags/timestamp_ingest_dag.py' 

# --- 2. EXECUTION ---

print(f"Attempting to upload file using profile: {PROFILE_NAME}")

try:
    # Create a session using the specified profile for authentication
    session = boto3.session.Session(profile_name=PROFILE_NAME)
    s3 = session.client('s3')

    # Use the dedicated upload_file method for easy transfer
    s3.upload_file(
        Filename=LOCAL_FILE_PATH,
        Bucket=BUCKET_NAME,
        Key=S3_KEY
    )

    print("-" * 40)
    print(f"✅ Success! File uploaded from:")
    print(f"  Local: {LOCAL_FILE_PATH}")
    print(f"  To S3: s3://{BUCKET_NAME}/{S3_KEY}")
    print("-" * 40)

except Exception as e:
    print(f"❌ An error occurred during S3 upload.")
    # This will often be due to incorrect profile name or IAM permissions
    print(f"   Error: {e}")
```
